# Violation Exploration

Notebook inicial para explorar os CSVs gerados por `src/identify_false_positives.py`.

```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
OUTPUT = BASE_DIR / 'output'

analysis_csv = OUTPUT / 'violations_unordered_57_cs_projects_with_pc_fp_analysis.csv'
all_violations_csv = OUTPUT / 'violations_unordered_57_cs_projects_with_pc_fp_analysis_violations_all.csv'
filtered_violations_csv = OUTPUT / 'violations_unordered_57_cs_projects_with_pc_fp_analysis_violations_filtered.csv'
context_csv = OUTPUT / 'violations_unordered_57_cs_projects_with_pc_fp_analysis_context_summary.csv'
coverage_csv = OUTPUT / 'violations_unordered_57_cs_projects_with_pc_fp_analysis_coverage_evidence.csv'

analysis_df = pd.read_csv(analysis_csv)
all_violations_df = pd.read_csv(all_violations_csv)
filtered_violations_df = pd.read_csv(filtered_violations_csv)
context_df = pd.read_csv(context_csv)
coverage_df = pd.read_csv(coverage_csv)
```

```python
summary = {
    'rows_total': len(analysis_df),
    'violations_before_filters': len(all_violations_df),
    'violations_after_filters': len(filtered_violations_df),
    'confirmed_false_positives': int((analysis_df['FPStatus'] == 'ConfirmedFalsePositive').sum()),
    'candidate_violations': int((analysis_df['FPStatus'] == 'CandidateViolation').sum()),
    'coverage_witnesses': len(coverage_df),
}
pd.Series(summary)
```
